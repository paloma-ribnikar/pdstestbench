## 🔢 1. ¿Qué significan los números `In [ ]` al lado de cada celda?

* **`In [ ]` (Casilla vacía):** Indica que la celda de código **aún no ha sido ejecutada**.
* **`In [*]` (Asterisco):** Indica que la celda **se está ejecutando en este momento** (Python está procesando el código). Si tarda mucho, el asterisco seguirá fijo.
* **`In [16]` (Número):** Indica el **orden exacto de ejecución**. Muestra el número de secuencia en el que corriste ese bloque durante tu sesión de trabajo.

## 🔄 2. Controles del Kernel (Memoria y Ejecución)

El **Kernel** es el motor silencioso de Python que corre por detrás y recuerda todas tus variables (como `x = 45`). En la barra de herramientas de arriba tenés los siguientes controles:

* **Interrupt (⏹️ Stop / Interrumpir):** Frena inmediatamente la ejecución de un código que se congeló o quedó colgado en un bucle infinito (sin borrar las variables guardadas).
* **Restart (🔄 Reiniciar Kernel):** Elimina todas las variables almacenadas en la memoria de Python (deja la memoria en cero), pero mantiene los gráficos y respuestas visibles en pantalla.
* **Restart & Clear Output:** Reinicia la memoria de Python **Y ADEMÁS borra todas las respuestas y gráficos** que se imprimieron abajo de las celdas, dejando el cuaderno 100% limpio como recién creado.
* **Restart & Run All:** Reinicia la memoria de Python, borra las salidas anteriores y **vuelve a ejecutar automáticamente todas las celdas en orden** desde la primera hasta la última.

## 🎹 3. Atajos de Teclado Esenciales

Jupyter tiene dos modos principales:
1. **Modo Edición (Celda en Verde/Azul activo):** Estás escribiendo dentro de la celda. (Presioná `Esc` para salir al Modo Comando).
2. **Modo Comando (Celda seleccionada sin cursor):** Estás seleccionando celdas enteras. (Presioná `Enter` para entrar al Modo Edición).

### 🚀 Ejecutar Celdas
* `Shift + Enter` : Ejecuta la celda actual y pasa a la siguiente (la más usada).
* `Ctrl + Enter`  : Ejecuta la celda actual y se queda parado en ella.
* `Alt + Enter`   : Ejecuta la celda actual e inserta una nueva celda abajo.

### ➕ Insertar, Borrar y Unir Celdas (En Modo Comando - apretar `Esc` primero)
* `A` : Inserta una celda arriba (*Above*).
* `B` : Inserta una celda abajo (*Below*).
* `D` + `D` (Doble D) : Borra la celda seleccionada (*Delete*).
* `Z` : Deshace la última eliminación de celda.
* `Shift + M` : **Une / Combina** dos o más celdas seleccionadas en una sola (*Merge*).

### 🔄 Cambiar el Tipo de Celda (En Modo Comando)
* `M` : Transforma la celda a **Markdown** (Texto).
* `Y` : Transforma la celda a **Code** (Código de Python).

## 🐍 4. Comandos Especiales ("Magics" de Jupyter)

Jupyter permite ejecutar comandos especiales antecedidos por un `%` (para Jupyter) o `!` (para la consola del sistema):

* `%timeit` : Mide el **tiempo de ejecución** exacto en milisegundos que tarda un código en ejecutarse (ideal para optimizar funciones en ingeniería).
* `%lsmagic` : Muestra en pantalla la **lista completa** de todos los comandos mágicos disponibles en Jupyter.
* `%matplotlib inline` : Hace que los gráficos de Matplotlib aparezcan incrustados prolijamente dentro del notebook.
* `!comando` : Ejecuta comandos de la terminal/consola directamente dentro de la celda de código (ej: `!dir` o `!pip install paquete`).
* `?nombre_funcion` : Muestra la ayuda y documentación oficial de cualquier función de Python (ej: `?np.linspace`).

### ✍️ Formato de Texto
* `*Texto en Cursiva*` ➔ *Texto en Cursiva*
* `**Texto en Negrita**` ➔ **Texto en Negrita**
* `***Negrita y Cursiva***` ➔ ***Negrita y Cursiva***
* `~~Texto Tachado~~` ➔ ~~Texto Tachado~~
* `` `código en línea` `` ➔ `código en línea`
* `<font color='blue'>Texto Azul</font>` ➔ <font color='blue'>Texto Azul</font>

### 🔗 Links vs Imágenes (Diferencia Clave)
* **Hipervínculo / Link (Sin signo `!`):**  
  Sintaxis: `[Texto que hace clic](https://github.com)`  
  Resultado: [Texto que hace clic](https://github.com)

* **Imagen (Empieza con signo `!`):**  
  Sintaxis: `![Descripción de la foto](URL_de_la_imagen)`  
  *(El signo `!` le avisa a Jupyter que debe mostrar la foto en lugar del texto del link)*.

## 📐 7. Ecuaciones Matemáticas con LaTeX

### 🔹 Ecuaciones en línea (dentro del párrafo)
Usá un solo signo de dólar `$` al inicio y al final:
* Sintaxis: `La frecuencia de muestreo es $f_s = \frac{1}{t_s}$.`
* Resultado: La frecuencia de muestreo es $f_s = \frac{1}{t_s}$.

### 🔹 Ecuaciones en bloque (centradas en su propia línea)
Usá doble signo de dólar `$$` al inicio y al final:

$$x(t) = A \cdot \sin(2\pi \cdot f \cdot t + \phi)$$

$$X(k) = \sum_{n=0}^{N-1} x(n) \cdot e^{-j 2\pi \frac{k n}{N}}$$

### 🔤 Símbolos Matemáticos Frecuentes en APS
| Símbolo | Sintaxis LaTeX |
| :--- | :--- |
| Letras Griegas | `\alpha`, `\beta`, `\pi`, `\sigma`, `\omega`, `\phi` |
| Fracción | `\frac{numerador}{denominador}` |
| Raíz Cuadrada | `\sqrt{x}` |
| Sumatoria | `\sum_{n=0}^{N-1}` |
| Integración | `\int_{-\infty}^{\infty}` |
| Estimador / Sombrero | `\hat{\sigma}^2` |
