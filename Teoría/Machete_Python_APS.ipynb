## 1️⃣ Definición de Funciones Básicas (`def`)

Las funciones te permiten empaquetar bloques de código reutilizables. En Python la sintaxis se define con la palabra clave `def` seguida del nombre de la función y paréntesis.

* **Indentación obligatoria:** Todo el código que pertenece a la función debe estar sangrado (4 espacios o 1 Tabulador).
* **Argumentos por defecto:** Podés definir valores por omisión (ej: `bb=0, cc=0`). Si al llamar a la función no le pasás esos parámetros, automáticamente toman el valor por defecto.

```python
# Función identidad básica
def una_funcion_que_no_hace_nada(a):
    return a

# Función que suma tres valores (bb y cc opcionales con valor 0 por defecto)
def funcion_que_suma(aa, bb=0, cc=0):
    return aa + bb + cc

# Función que concatena textos
def funcion_que_concatena(s1, s2=''):
    return s1 + s2

# Pruebas de llamada a funciones:
print("Suma 1 + 2 + 0:", funcion_que_suma(1, 2))
print("Suma 1 + 0 + 2 (especificando cc):", funcion_que_suma(1, cc=2))
print("Concatenar textos:", funcion_que_concatena("Hola ", "Mundo"))
```

## 2️⃣ Uso de Variables, Cadenas, Listas e Inspección (`isinstance`)

Python maneja dinámicamente el tipo de datos numéricos (`int` y `float`) y textos (`str`). Las **listas** son colecciones ordenadas e iterables que pueden guardar elementos de distinto tipo.

### 📌 Reglas de Indexación, Slicing e Inspección:
* `lista[0]` : Primer elemento (índice 0).
* `lista[-1]` : Último elemento (índice negativo cuenta desde el final).
* `lista[inicio:fin]` : Extrae el subrango desde `inicio` hasta `fin - 1` (Slicing).
* `lista[inicio:fin:paso]` : Salta de a `paso` elementos (ej: `[0::2]` toma los elementos pares).
* `lista[::-1]` : Invierte completamente la lista.
* `isinstance(variable, tipo)` : Verifica si una variable pertenece a un tipo de dato (ej: `numbers.Number` o `str`).

```python
import numbers

# Variables numéricas y de texto
Fs = 1000.0  # Frecuencia de muestreo (Hz)
N = 1000     # Cantidad de muestras

un_texto = 'señal 0'
otro_texto = "señal 1"
mi_texto = un_texto + otro_texto  # Concatenación de textos

# Construcción de una lista iterable
mi_lista = []
mi_lista += un_texto        # Agrega cada carácter de un_texto como elemento
mi_lista += [otro_texto]    # Agrega el texto entero como 1 elemento
mi_lista += [Fs]            # Agrega el flotante Fs

print("Lista completa:", mi_lista)
print("Primer elemento [0]:", mi_lista[0])
print("Último elemento [-1]:", mi_lista[-1])
print("Anteúltimo [-2]:", mi_lista[-2])

print("Rangos [0:3]:", mi_lista[0:3])
print("Elementos en posiciones pares [0::2]:", mi_lista[0::2])
print("Elementos en posiciones impares [1::2]:", mi_lista[1::2])
print("Lista invertida [::-1]:", mi_lista[::-1])

# Inspección con isinstance()
es_numero = [isinstance(aa, numbers.Number) for aa in mi_lista]
es_texto = [isinstance(aa, str) for aa in mi_lista]
print("¿Es número?:", es_numero)
print("¿Es texto?:", es_texto)
```

## 3️⃣ NumPy: Secuencias, Rangos y Grillas Temporales (`np.arange` vs `np.linspace`)

El módulo `numpy` (abreviado `np`) agrega los tipos numéricos y métodos del álgebra matricial para Procesamiento Digital de Señales.

### 🔹 Métodos de Creación de Vectores:
* `np.arange(inicio, fin, paso)` : Genera valores espaciados de a `paso` (similar a la sintaxis `inicio:paso:fin` de MATLAB).
* `np.linspace(inicio, fin, cantidad_puntos)` : Genera exactamente `cantidad_puntos` equidistantes entre `inicio` y `fin`. **Es el método estándar recomendado en los testbenchs de la materia**.

### 📐 Conceptos de PDS:
* **`fs` ($f_s$):** Frecuencia de muestreo (muestras por segundo).
* **`ts` ($t_s = 1/f_s$):** Período de muestreo (tiempo entre muestra y muestra).
* **`N`:** Cantidad total de puntos o muestras procesadas.
* **Frecuencia de Nyquist ($f_{nyq} = f_s / 2$):** Máxima frecuencia representable sin solapamiento (aliasing).

```python
import numpy as np

# Secuencia de enteros y flotantes con arange
vector_int = np.arange(10)
vector_float = np.arange(5.0, 10.0, 0.1)

N = 1000
Fs = 1000.0

# Grilla temporal con np.arange
vector_tiempo_arange = np.arange(0.0, N/Fs, 1/Fs)

# Grilla frecuencial desde 0 Hz hasta Nyquist (Fs/2)
vector_frecuencia = np.arange(0.0, Fs/2.0, N/Fs)

# Grilla temporal recomendada en testbenchs con np.linspace
ts = 1/Fs
tt = np.linspace(0, (N-1)*ts, N).flatten()

print("Total muestras en tt:", len(tt))
print("Primeras 5 muestras de tiempo [s]:", tt[:5])
```

## 4️⃣ Matrices de Canales ADC, Ruido Blanco y Filtrado Booleano (`np.nonzero`)

Podemos simular la lectura multicanal de un Conversor Analógico-Digital (ADC) utilizando matrices aleatorias de NumPy.

### 📌 Direccionamiento Booleano e Índices:
* `np.random.normal(media, desvio, size=[filas, col])` : Genera datos aleatorios con distribución Gaussiana.
* `bool_vector = abs(matriz[:, 0]) > 0.5` : Crea una máscara lógica (True/False) basada en una condición.
* `np.nonzero(bool_vector)[0]` : Devuelve los índices exactos donde la condición booleana es verdadera (desempaquetando la tupla).
* `matriz[bool_vector, :]` : Filtra la matriz extrayendo solo las filas que cumplen la condición.

```python
canales_ADC = 8
# Simular 8 canales ADC con ruido normal distribuido N(0, 1)
matriz_datos = np.random.normal(0, 1.0, size=[N, canales_ADC])

# Máscara booleana: buscar donde la amplitud absoluta del canal 0 supere 0.5 V
bool_vector = abs(matriz_datos[:, 0]) > 0.5
indices = np.nonzero(bool_vector)[0]

print("Cantidad de muestras que superan 0.5 V:", len(indices))
print("Filas filtradas por la condición booleana:", matriz_datos[bool_vector, :].shape)
```

## 6️⃣ Estructuras de Control e Iteraciones (`for`, `range`, `zip`)

En Python existen distintas formas de iterar sobre colecciones de datos:

* `for elemento in lista:` : Itera directamente sobre los valores de la lista (sintaxis Pythonica preferida).
* `for i in range(len(lista)):` : Itera mediante un índice numérico de 0 a $N-1$ (estilo C / MATLAB).
* `for a, b in zip(lista1, lista2):` : Empaqueta dos o más listas y las recorre en paralelo simultáneamente.

```python
print("--- Iteración directa por elementos ---")
for aa in mi_lista:
    print(aa)

print("\n--- Iteración mediante índices range() ---")
for ii in range(len(mi_lista)):
    print(f"Índice {ii}: {mi_lista[ii]}")

print("\n--- Iteración en paralelo con zip() ---")
for aa, bb in zip(mi_lista, mi_lista[::-1]):
    print(f"{aa} <--> {bb}")
```

## 7️⃣ Generación de Señales Senoidales, Ruido Gaussiano y Diccionarios de Configuración

Esta es la sintaxis utilizada en los **Testbenchs** para simular señales y ruido en la materia:

### 🌊 Fórmula de Señal Senoidal:
$$s(t) = A \cdot \sin(2\pi \cdot f \cdot t + \phi)$$
* `A`: Amplitud en Voltios.
* `f`: Frecuencia en Hertz (Hz).
* `phi`: Fase inicial en radianes.

### 🎲 Ruido Gaussiano y Varianza ($\sigma^2$):
* `ruido = np.sqrt(varianza) * np.random.randn(N)` : Genera ruido aleatorio blanco. La **varianza** ($\sigma^2$) representa la **potencia/fuerza del ruido**.

```python
# Configuración de señal utilizando Diccionarios (Sintaxis oficial de Testbench)
sig_props = {
    'tipo': 'senoidal',
    'frecuencia': (3, 10, 20), # Tupla con frecuencias en Hz
    'amplitud':   (1, 1, 1),   # Tupla con amplitudes en V
    'fase':       (0, 0, 0)    # Tupla con fases
}
# Generar etiquetas de descripción de forma programática
sig_props['descripcion'] = [str(f) + ' Hz' for f in sig_props['frecuencia']]

# Generación matemática de una Senoidal pura (10 Hz, Amplitud 2V)
frecuencia_seno = 10.0
amplitud_seno = 2.0
seno = amplitud_seno * np.sin(2 * np.pi * frecuencia_seno * tt)

# Generación de Ruido Gaussiano blanco con varianza 1.0
varianza_ruido = 1.0
ruido = np.sqrt(varianza_ruido) * np.random.randn(N)

print("Etiquetas generadas:", sig_props['descripcion'])
```

## 8️⃣ Graficación Profesional con Matplotlib (`plt`)

Para visualizar las señales en función del tiempo utilizamos `matplotlib.pyplot`:

* `plt.figure(figsize=(anchura, altura))` : Crea el marco gráfico.
* `plt.plot(eje_x, eje_y, label='Etiqueta')` : Dibuja la curva cartesiana.
* `plt.title()`, `plt.xlabel()`, `plt.ylabel()` : Agrega títulos y nombres a los ejes.
* `plt.legend(loc='upper right')` : Muestra el recuadro con la leyenda explicativa.
* `plt.grid(True)` : Activa la grilla cuadriculada de fondo.
* `plt.show()` : Renderiza y muestra el gráfico final en pantalla.

```python
import matplotlib.pyplot as plt

# Crear marco del gráfico
plt.figure(figsize=(10, 4))

# Plotear Senoidal Pura vs Senoidal con Ruido
plt.plot(tt, seno, label='Senoidal Pura (10 Hz)', color='blue', linewidth=1.5)
plt.plot(tt, seno + ruido, label='Senoidal + Ruido Gaussiano', color='orange', alpha=0.6)

# Formatear ejes y etiquetas
plt.title('Simulación de Señal y Ruido en el Tiempo (APS)')
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Amplitud [V]')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
```
